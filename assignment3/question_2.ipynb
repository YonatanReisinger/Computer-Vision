# Question 2 — Rotation and Translation Transformations

## Setup

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import math

img = cv2.imread('tel_aviv_is_red.jpg', cv2.IMREAD_COLOR_RGB)

def show_images(images, titles, rows, cols, figsize=(15, 10)):
    fig, axes = plt.subplots(rows, cols, figsize=figsize, squeeze=False)
    axes = axes.flatten()

    for img, title, ax in zip(images, titles, axes):
        ax.imshow(img, cmap='gray' if len(img.shape) == 2 else None)
        ax.set_title(title)
        ax.axis('off')

    plt.tight_layout()
    plt.show()

show_images([img], ['Tel Aviv is Red'], 1, 1)
```

## Rotation

### 1.1 Rotation Using OpenCV Built-in Functions

```python
h, w = img.shape[:2]
center = (w // 2, h // 2)

angles = [30, 45, 90]
interpolations = {
    "Nearest Neighbor": cv2.INTER_NEAREST,
    "Bilinear": cv2.INTER_LINEAR,
    "Bicubic": cv2.INTER_CUBIC
}

results = []
titles = []

for angle in angles:
    M = cv2.getRotationMatrix2D(center, angle, 1.0)

    for interp_name, interp_flag in interpolations.items():
        rotated_img = cv2.warpAffine(img, M, (w, h), flags=interp_flag)
        results.append(rotated_img)
        titles.append(f"Angle: {angle}, {interp_name}")

show_images(results, titles, 3, 3)
```

### 1.2 Rotation Using Your Own Implementation

```python
def manual_rotation(image, angle):
    theta = math.radians(angle)
    h, w = image.shape[:2]

    out_img = np.zeros_like(image)

    cx, cy = w // 2, h // 2

    cos_theta = math.cos(theta)
    sin_theta = math.sin(theta)

    for y in range(h):
        for x in range(w):

            x_shifted = x - cx
            y_shifted = y - cy

            src_x = int(round(x_shifted * cos_theta - y_shifted * sin_theta)) + cx
            src_y = int(round(x_shifted * sin_theta + y_shifted * cos_theta)) + cy

            if 0 <= src_x < w and 0 <= src_y < h:
                out_img[y, x] = image[src_y, src_x]

    return out_img

angles = [30, 45, 90]
manual_results = [manual_rotation(img, angle) for angle in angles]
manual_titles = [f"Manual Rotation: {angle}" for angle in angles]

show_images(manual_results, manual_titles, 1, 3, figsize=(15, 5))
```

### Comparison

```python
# Define angles
angles = [30, 45, 90] # [cite: 129, 130, 131]

h, w = img.shape[:2]
center = (w // 2, h // 2)

cv2_images = []
cv2_titles = []
manual_images = []
manual_titles = []

for angle in angles:
    M = cv2.getRotationMatrix2D(center, angle, 1.0)
    cv2_rotated = cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_NEAREST)
    cv2_images.append(cv2_rotated)
    cv2_titles.append(f"OpenCV: {angle}°")

    manual_rotated = manual_rotation(img, angle)
    manual_images.append(manual_rotated)
    manual_titles.append(f"Manual: {angle}°")

comparison_images = cv2_images + manual_images
comparison_titles = cv2_titles + manual_titles

print("Side-by-Side Comparison: OpenCV vs. Manual Implementation")
show_images(comparison_images, comparison_titles, rows=2, cols=3, figsize=(15, 10))
```

## Translation

### 2.1 Translation Using OpenCV

```python
translations = [
    (50, 0),
    (0, 50),
    (50, 50),
    (-50, -50)
]

cv2_trans_results = []
cv2_trans_titles = []

for tx, ty in translations:
    M = np.float32([[1, 0, tx],
                    [0, 1, ty]])

    translated_img = cv2.warpAffine(img, M, (w, h))
    cv2_trans_results.append(translated_img)
    cv2_trans_titles.append(f"tx={tx}, ty={ty}")

show_images(cv2_trans_results, cv2_trans_titles, 2, 2)
```

### 2.2 Translation Using Your Own Implementation

```python
def manual_translation(image, tx, ty):
    h, w = image.shape[:2]
    out_img = np.zeros_like(image)

    for y in range(h):
        for x in range(w):
            new_x = x + tx
            new_y = y + ty

            if 0 <= new_x < w and 0 <= new_y < h:
                out_img[new_y, new_x] = image[y, x]

    return out_img

manual_trans_results = [manual_translation(img, tx, ty) for tx, ty in translations]
manual_trans_titles = [f"Manual tx={tx}, ty={ty}" for tx, ty in translations]

show_images(manual_trans_results, manual_trans_titles, 2, 2)
```
